```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('amazon.csv')
```

```python
df.head(10)
```

```python
df.info()
```

```python
cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']

for col in cols:
    df[col] = df[col].astype(str).str.strip()

for col in ['discounted_price', 'actual_price']:
    df[col] = df[col].str.replace('₹', '', regex=False).str.replace(',', '', regex=False)

df['discount_percentage'] = df['discount_percentage'].str.replace('%', '', regex=False)
df['rating_count'] = df['rating_count'].str.replace(',', '', regex=False)

df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['rating_count'])
df = df.dropna(subset=['rating'])
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates(subset='product_id')
```

```python
df.shape
```

```python
df['category'] = df['category'].str.split('|').str[1]
```

```python
df.head(10)
```

```python
df[cols].describe()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df['actual_price'], bins=10)
plt.title('Actual Price')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df['rating'], bins=10)
plt.title('Rating')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df['rating_count'], bins=10)
plt.title('Rating Count')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df['discount_percentage'], bins=10)
plt.title('Discount Percentage')
plt.show()
```

```python
top_cat = df['category'].value_counts().head(10).sort_values()

plt.figure(figsize=(10, 5))
plt.barh(top_cat.index, top_cat.values)
plt.title('Top 10 Categories')
plt.show()
```

```python
cat_rating = df.groupby('category')['rating'].mean().sort_values()

plt.figure(figsize=(8, 8))
cat_rating.plot(kind='barh')
plt.title('Average Rating by Category')
plt.show()
```

```python
cat_discount = df.groupby('category')['discount_percentage'].mean()

plt.figure(figsize=(12, 6))
cat_discount.plot(kind='bar')
plt.xticks(rotation=80)
plt.title('Average Discount by Category')
plt.show()
```

```python
rate_plot = df.groupby('category').agg({'rating': 'mean', 'rating_count': 'sum'}).sort_values('rating_count', ascending=False).reset_index()
rate_plot.head()
```

```python
fig, ax1 = plt.subplots(figsize=(14, 7))
ax1.bar(rate_plot['category'], rate_plot['rating_count'])
ax2 = ax1.twinx()
ax2.scatter(range(len(rate_plot)), rate_plot['rating'], color='red', s=60)
ax2.set_ylabel('Average Rating')
ax1.tick_params(axis='x', labelrotation=90)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['actual_price'], df['rating'])
plt.xlabel('actual_price')
plt.ylabel('rating')
plt.title('Actual Price vs Rating')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['actual_price'], df['discounted_price'])
plt.xlabel('actual_price')
plt.ylabel('discounted_price')
plt.title('Actual Price vs Discounted Price')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['rating_count'], df['rating'])
plt.xlabel('rating_count')
plt.ylabel('rating')
plt.title('Rating Count vs Rating')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['discount_percentage'], df['rating'])
plt.xlabel('discount_percentage')
plt.ylabel('rating')
plt.title('Discount vs Rating')
plt.show()
```
